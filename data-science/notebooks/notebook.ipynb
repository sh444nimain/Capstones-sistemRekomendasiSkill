### Cleaning Data

```python
all_job_df['job_title'] = all_job_df['job_title'].str.lower().str.strip()
all_job_df['job_description'] = all_job_df['job_description'].str.lower().str.strip()
all_job_df['job_skill_set'] = all_job_df['job_skill_set'].str.lower().str.strip()
all_job_df['category'] = all_job_df['category'].str.lower().str.strip()
```

Standardisasi teks

```python
cols = ['job_title', 'job_description', 'job_skill_set', 'category']

for col in cols:
    all_job_df[col] = all_job_df[col].str.replace(r'\s+', ' ', regex=True)
```

Membersihkan double spasi pada setiap kolom

```python
all_job_df['category'] = all_job_df['category'].str.replace('-', ' ')
```

Merapihkan format category

```python
all_job_df['job_description'] = all_job_df['job_description'].apply(
    lambda x: re.sub(r'[^a-zA-Z0-9\s]', '', x.replace('\n', ' '))
)
```
